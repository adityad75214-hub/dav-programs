```python
# ============================================================
# INTRODUCTION TO PYTHON FOR DATA ANALYTICS
# NumPy + Pandas
# ============================================================

import numpy as np
import pandas as pd


# ============================================================
# 1. NUMPY ARRAY CREATION
# ============================================================

print("========== NUMPY ARRAY CREATION ==========")

arr1 = np.array([10, 20, 30, 40, 50])

arr2 = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("1D Array:")
print(arr1)

print("\n2D Array:")
print(arr2)

print("\nType:", type(arr1))
print("Data Type:", arr1.dtype)
print("Shape of 1D array:", arr1.shape)
print("Shape of 2D array:", arr2.shape)


# ============================================================
# 2. NUMPY ARRAY OPERATIONS
# ============================================================

print("\n========== ARRAY OPERATIONS ==========")

a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

print("Array A:", a)
print("Array B:", b)

print("\nAddition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)


# ============================================================
# 3. SCALAR OPERATIONS
# ============================================================

print("\n========== SCALAR OPERATIONS ==========")

arr = np.array([10, 20, 30, 40])

print("Original:", arr)
print("Add 5:", arr + 5)
print("Subtract 5:", arr - 5)
print("Multiply by 2:", arr * 2)
print("Divide by 2:", arr / 2)


# ============================================================
# 4. NUMPY INDEXING
# ============================================================

print("\n========== NUMPY INDEXING ==========")

arr = np.array([10, 20, 30, 40, 50])

print("Array:", arr)

print("First element:", arr[0])
print("Third element:", arr[2])
print("Last element:", arr[-1])


# ============================================================
# 5. NUMPY SLICING
# ============================================================

print("\n========== NUMPY SLICING ==========")

print("Elements from index 1 to 3:", arr[1:4])
print("First three elements:", arr[:3])
print("From index 2 onwards:", arr[2:])
print("Every second element:", arr[::2])


# ============================================================
# 6. 2D ARRAY INDEXING
# ============================================================

print("\n========== 2D ARRAY INDEXING ==========")

matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Matrix:")
print(matrix)

print("\nElement at row 0, column 0:", matrix[0, 0])
print("Element at row 1, column 2:", matrix[1, 2])
print("Element at row 2, column 1:", matrix[2, 1])


# ============================================================
# 7. 2D ARRAY SLICING
# ============================================================

print("\n========== 2D ARRAY SLICING ==========")

print("First two rows and first two columns:")
print(matrix[0:2, 0:2])

print("\nSecond column:")
print(matrix[:, 1])

print("\nSecond row:")
print(matrix[1, :])


# ============================================================
# 8. PANDAS DATAFRAME CREATION
# ============================================================

print("\n========== PANDAS DATAFRAME ==========")

data = {
    "Name": ["Aditya", "Rahul", "Aman", "Rohit"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 78, 92, 88]
}

df = pd.DataFrame(data)

print(df)


# ============================================================
# 9. DATAFRAME INFORMATION
# ============================================================

print("\n========== DATAFRAME INFORMATION ==========")

print("Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)


# ============================================================
# 10. PANDAS COLUMN INDEXING
# ============================================================

print("\n========== COLUMN INDEXING ==========")

print("Name column:")
print(df["Name"])

print("\nMarks column:")
print(df["Marks"])

print("\nName and Marks:")
print(df[["Name", "Marks"]])


# ============================================================
# 11. PANDAS ROW INDEXING USING iloc
# ============================================================

print("\n========== ROW INDEXING ==========")

print("First row:")
print(df.iloc[0])

print("\nSecond row:")
print(df.iloc[1])


# ============================================================
# 12. PANDAS ROW SLICING
# ============================================================

print("\n========== ROW SLICING ==========")

print("First two rows:")
print(df.iloc[0:2])

print("\nRows 2 to 4:")
print(df.iloc[1:4])


# ============================================================
# 13. ROW + COLUMN SLICING
# ============================================================

print("\n========== ROW + COLUMN SLICING ==========")

print("First three rows and first two columns:")
print(df.iloc[0:3, 0:2])


# ============================================================
# 14. COMPLETE DATAFRAME
# ============================================================

print("\n========== COMPLETE DATAFRAME ==========")

print(df)
```

```python
# ============================================================
# EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================

import pandas as pd


# ============================================================
# 1. CREATE / LOAD DATASET
# ============================================================

data = {
    "Name": [
        "Aditya",
        "Rahul",
        "Aman",
        "Rohit",
        "Priya",
        "Aditya"
    ],

    "Age": [
        20,
        21,
        19,
        None,
        22,
        20
    ],

    "Marks": [
        85,
        78,
        92,
        88,
        None,
        85
    ],

    "Branch": [
        "CSE",
        "ECE",
        "CSE",
        "ISE",
        "CSE",
        "CSE"
    ]
}

df = pd.DataFrame(data)


# ============================================================
# 2. DISPLAY DATASET
# ============================================================

print("========== DATASET ==========")
print(df)


# ============================================================
# 3. FIRST FEW RECORDS
# ============================================================

print("\n========== FIRST 5 RECORDS ==========")
print(df.head())


# ============================================================
# 4. LAST FEW RECORDS
# ============================================================

print("\n========== LAST 5 RECORDS ==========")
print(df.tail())


# ============================================================
# 5. BASIC INFORMATION
# ============================================================

print("\n========== DATASET INFORMATION ==========")

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nComplete Information:")
df.info()


# ============================================================
# 6. DESCRIPTIVE STATISTICS
# ============================================================

print("\n========== DESCRIPTIVE STATISTICS ==========")

print(df.describe())


# ============================================================
# 7. INDIVIDUAL STATISTICS
# ============================================================

print("\n========== INDIVIDUAL STATISTICS ==========")

print("Mean Marks:", df["Marks"].mean())
print("Median Marks:", df["Marks"].median())
print("Minimum Marks:", df["Marks"].min())
print("Maximum Marks:", df["Marks"].max())
print("Standard Deviation:", df["Marks"].std())


# ============================================================
# 8. CHECK MISSING VALUES
# ============================================================

print("\n========== MISSING VALUES ==========")

print(df.isnull())


# ============================================================
# 9. COUNT MISSING VALUES
# ============================================================

print("\n========== NUMBER OF MISSING VALUES ==========")

print(df.isnull().sum())


# ============================================================
# 10. CHECK WHETHER DATASET HAS ANY MISSING VALUE
# ============================================================

print("\n========== ANY MISSING VALUES? ==========")

print(df.isnull().any())


# ============================================================
# 11. FIND ROWS CONTAINING MISSING VALUES
# ============================================================

print("\n========== ROWS WITH MISSING VALUES ==========")

print(df[df.isnull().any(axis=1)])


# ============================================================
# 12. CHECK DUPLICATES
# ============================================================

print("\n========== DUPLICATE RECORDS ==========")

print(df.duplicated())


# ============================================================
# 13. COUNT DUPLICATES
# ============================================================

print("\n========== NUMBER OF DUPLICATES ==========")

print(df.duplicated().sum())


# ============================================================
# 14. DISPLAY DUPLICATE ROWS
# ============================================================

print("\n========== DUPLICATE ROWS ==========")

print(df[df.duplicated()])


# ============================================================
# 15. REMOVE DUPLICATES
# ============================================================

df_no_duplicates = df.drop_duplicates()

print("\n========== DATA AFTER REMOVING DUPLICATES ==========")

print(df_no_duplicates)


# ============================================================
# 16. HANDLE MISSING VALUES
# ============================================================

# Fill missing numerical values with their mean

df_clean = df_no_duplicates.copy()

df_clean["Age"] = df_clean["Age"].fillna(df_clean["Age"].mean())
df_clean["Marks"] = df_clean["Marks"].fillna(df_clean["Marks"].mean())


# ============================================================
# 17. FINAL CLEAN DATASET
# ============================================================

print("\n========== CLEAN DATASET ==========")

print(df_clean)


# ============================================================
# 18. CHECK AGAIN
# ============================================================

print("\n========== FINAL CHECK ==========")

print("Missing values:")
print(df_clean.isnull().sum())

print("\nDuplicate rows:")
print(df_clean.duplicated().sum())
```

```python
# ============================================================
# DATA VISUALIZATION USING MATPLOTLIB AND SEABORN
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 1. CREATE DATASET
# ============================================================

data = {
    "Name": ["Aditya", "Rahul", "Aman", "Rohit", "Priya",
             "Karan", "Neha", "Ankit", "Pooja", "Vikas"],

    "Age": [20, 21, 19, 22, 20, 23, 21, 20, 22, 19],

    "StudyHours": [5, 6, 3, 8, 4, 9, 7, 5, 8, 2],

    "Marks": [75, 82, 60, 90, 70, 95, 85, 78, 88, 55]
}

df = pd.DataFrame(data)

print("========== DATASET ==========")
print(df)


# ============================================================
# 2. HISTOGRAM
# ============================================================

plt.figure()

plt.hist(df["Marks"], bins=5)

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()


# ============================================================
# 3. BOX PLOT
# ============================================================

plt.figure()

sns.boxplot(x=df["Marks"])

plt.title("Box Plot of Marks")
plt.xlabel("Marks")

plt.show()


# ============================================================
# 4. SCATTER PLOT
# ============================================================

plt.figure()

plt.scatter(df["StudyHours"], df["Marks"])

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()


# ============================================================
# 5. SCATTER PLOT USING SEABORN
# ============================================================

sns.scatterplot(
    data=df,
    x="StudyHours",
    y="Marks"
)

plt.title("Study Hours vs Marks")

plt.show()


# ============================================================
# 6. PAIR PLOT
# ============================================================

sns.pairplot(
    df[["Age", "StudyHours", "Marks"]]
)

plt.show()


# ============================================================
# 7. CORRELATION MATRIX
# ============================================================

correlation = df[
    ["Age", "StudyHours", "Marks"]
].corr()

print("\n========== CORRELATION MATRIX ==========")

print(correlation)


# ============================================================
# 8. CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True
)

plt.title("Correlation Heatmap")

plt.show()
```
